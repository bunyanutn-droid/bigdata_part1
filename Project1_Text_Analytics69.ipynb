```python
# จำนวนบรรทัดสูงสุดที่จะอ่านต่อไฟล์ (กันไม่ให้ช้าเกินไปตอนรันสำรวจข้อมูล)
# ปรับเพิ่มได้ถ้าต้องการตัวอย่างใหญ่ขึ้น
MAX_LINES_PER_FILE = 20_000

# จำนวนไฟล์ที่จะใช้ (โจทย์บอกว่าใช้ 1-3 ไฟล์ก็พอ ไม่ต้องโหลดทั้ง dataset)
N_FILES = 2
```

```python
# ---------- 1) เลือกไฟล์ .jsonl ต้นฉบับจาก repo ----------
def pick_files(repo_id: str, n_files: int):
    """ดึงรายชื่อไฟล์ .jsonl ทั้งหมดใน repo แล้วเลือกมาใช้ n_files ไฟล์
    (เลือกแบบกระจายช่วงเวลา เพื่อให้เห็นหลายชั่วโมง)"""
    all_files = list_repo_files(repo_id, repo_type="dataset")
    jsonl_files = sorted(f for f in all_files if f.endswith(".jsonl"))
    if not jsonl_files:
        raise RuntimeError("ไม่พบไฟล์ .jsonl ใน repo นี้ ตรวจสอบชื่อ dataset อีกครั้ง")

    if n_files >= len(jsonl_files):
        return jsonl_files

    # กระจายการเลือกไฟล์ให้ครอบคลุมช่วงเวลาที่กว้างขึ้น (เช่น เช้า/บ่าย)
    step = len(jsonl_files) / n_files
    idx = [int(i * step) for i in range(n_files)]
    return [jsonl_files[i] for i in idx]
    # ---------- 2) generator อ่าน jsonl ทีละบรรทัด (ตามที่โจทย์แนะนำ) ----------
def iter_jsonl(path, limit=None):
    with open(path, encoding="utf-8") as f:
        for i, line in enumerate(f):
            if limit is not None and i >= limit:
                break
            line = line.strip()
            if not line:
                continue
            try:
                yield json.loads(line)
            except json.JSONDecodeError:
                continue


def detect_lang(text: str) -> str:
    text = text.strip()
    if len(text) < 3:
        return "(too_short/empty)"
    try:
        return detect(text)
    except Exception:
        return "unknown"


def main():
    global total_posts, min_time, max_time, hour_counter, lang_counter, hashtag_counter, img_count, reply_count, example_texts

    files_to_use = pick_files(REPO_ID, N_FILES)
    print(f"ใช้ไฟล์ทั้งหมด {len(files_to_use)} ไฟล์:")
    for f in files_to_use:
        print("  -", f)

    local_paths = [
        hf_hub_download(repo_id=REPO_ID, filename=f, repo_type="dataset")
        for f in files_to_use
    ]

    # ---------- ตัวนับต่างๆ (สร้างครั้งเดียว รวมทุกไฟล์) ----------
    total_posts = 0
    min_time, max_time = None, None
    hour_counter = collections.Counter()      # โพสต์ต่อ "ชั่วโมง" (YYYY-MM-DD HH:00)
    lang_counter = collections.Counter()
    hashtag_counter = collections.Counter()
    img_count = 0
    reply_count = 0
    example_texts = collections.defaultdict(list)  # เก็บตัวอย่างจริงต่อ hashtag เด่นๆ

    for path in local_paths:
        for post in iter_jsonl(path, limit=MAX_LINES_PER_FILE):
            total_posts += 1
            text = post.get("text", "") or ""

            # --- เวลา ---
            try:
                t = datetime.fromisoformat(post["created_at"].replace("Z", "+00:00"))
            except Exception:
                t = None
            if t is not None:
                if min_time is None or t < min_time:
                    min_time = t
                if max_time is None or t > max_time:
                    max_time = t
                hour_counter[t.strftime("%Y-%m-%d %H:00")] += 1

            # --- ภาษา ---
            lang_counter[detect_lang(text)] += 1

            # --- hashtag ---
            for tag in re.findall(r"#(\w+)", text):
                tag = tag.lower()
                hashtag_counter[tag] += 1
                if len(example_texts[tag]) < 2 and text.strip():
                    example_texts[tag].append(text.strip()[:120])

            # --- อื่นๆ ---
            if post.get("has_images"):
                img_count += 1
            if post.get("reply_to"):
                reply_count += 1


main()
```
